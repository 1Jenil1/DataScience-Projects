```python
# convert object to date and time formate
df['reservation_status_date'] = pd.to_datetime(df['reservation_status_date'], dayfirst=True, errors='coerce')
```

```python
df.describe(include = 'object')
```

# find missing value in columns 

```python
missing_values = df.isnull().sum()
print(missing_values)
```

# drope missing value columns & raw

```python
cols_to_drop = ['company', 'agent']
df = df.drop(columns=[col for col in cols_to_drop if col in df.columns], errors='ignore')
df.dropna(inplace=True)
```

# data analysis and visualization

```python
# find cancelled per for resrvition
cancelled_per = df['is_canceled'].value_counts(normalize = True)

plt.figure(figsize=(5, 4))
plt.title('Reservation Status Count')
plt.bar(['Not Canceled', 'Canceled'], df['is_canceled'].value_counts(), width=0.5)  # Fix function call
plt.xlabel("Status")
plt.ylabel("Count")
plt.show()
```

```python
# find diffrent between tow hotel booking cancelation.
plt.figure(figsize=(8,4))
ax1 = sns.countplot(x='hotel', hue='is_canceled', data=df, palette='Blues')
ax1.legend(title="Cancellation Status", bbox_to_anchor=(1, 1))
plt.title('Reservation Status in Different Hotels', size=20)
plt.xlabel('Hotel')
plt.ylabel('Number of Reservations')

plt.show()
```

```python
# find cancelation percentage of resort hotel
resort_hotel = df[df['hotel'] == 'Resort Hotel'] 
resort_hotel['is_canceled'].value_counts(normalize=True)
```

```python
# find cancelation percentage of city hotel
city_hotel = df[df['hotel'] == 'City Hotel']
city_hotel['is_canceled'].value_counts(normalize=True)
```

```python
resort_hotel = resort_hotel.groupby('reservation_status_date')[['adr']].mean()
city_hotel = city_hotel.groupby('reservation_status_date')[['adr']].mean()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(20, 8))
plt.title('Average Daily Rate in City and Resort Hotel', fontsize=30)
plt.plot(resort_hotel.index, resort_hotel['adr'], label='Resort Hotel')
plt.plot(city_hotel.index, city_hotel['adr'], label='City Hotel')
plt.legend(fontsize=20)
plt.xlabel('Date')
plt.ylabel('ADR (Average Daily Rate)')
plt.grid(True)
plt.show()
```

```python
# find per month Reseverstion
df['month'] = df['reservation_status_date'].dt.month
plt.figure(figsize = (16,8))
ax1 = sns.countplot(x='month',hue='is_canceled',data = df,palette = 'bright')
legend_labels,_ =ax1.get_legend_handles_labels()
ax1.legend(bbox_to_anchor=(1,1))
plt.title('Resevertion status per month',size = 20)
plt.xlabel('month')
plt.ylabel('number of reseversition')
plt.legend(['not canceled','canceled'])
plt.show()
```

```python
# find price of rent 

plt.figure(figsize=(15, 8))
plt.title('ADR per Month', fontsize=30)

# Group by 'arrival_date_month' and calculate the sum of 'adr'
adr_per_month = df[df['is_canceled'] == 1].groupby('arrival_date_month')[['adr']].sum().reset_index()

# Plot the results
sns.barplot(x='arrival_date_month', y='adr', data=adr_per_month)
plt.show()
```

```python
# find top 10 country cancellation 
cancelled_data = df[df['is_canceled']==1]
top_10_country = cancelled_data['country'].value_counts()[:10]
plt.figure(figsize = (8,8))
plt.title('Top 10 Countries with Reservation Cancellations')
plt.pie(top_10_country,autopct = '%.2f%%',labels = top_10_country.index)
plt.show()
```

```python
# Filter the data for cancelled and not cancelled
cancelled_data = df[df['is_canceled'] == 1]
not_cancelled_data = df[df['is_canceled'] == 0]

# Group and calculate mean ADR
cancelled_df_adr = cancelled_data.groupby('reservation_status_date')[['adr']].mean().reset_index()
cancelled_df_adr.sort_values('reservation_status_date', inplace=True)

not_cancelled_df_adr = not_cancelled_data.groupby('reservation_status_date')[['adr']].mean().reset_index()
not_cancelled_df_adr.sort_values('reservation_status_date', inplace=True)

# Plot the data
plt.figure(figsize=(20, 6))
plt.title('Average Daily Rate')
plt.plot(not_cancelled_df_adr['reservation_status_date'], not_cancelled_df_adr['adr'], label='Not Cancelled')
plt.plot(cancelled_df_adr['reservation_status_date'], cancelled_df_adr['adr'], label='Cancelled')
plt.legend()
plt.show()

```

```python
# select some specific year data 
# for filter cancellel data
cancelled_df_adr = cancelled_df_adr[(cancelled_df_adr['reservation_status_date'] > '2016-01-01') & 
                                     (cancelled_df_adr['reservation_status_date'] < '2017-09-01')]

# for filter the not cancelled data
not_cancelled_df_adr = not_cancelled_df_adr[(not_cancelled_df_adr['reservation_status_date'] > '2016-01-01') & 
                                             (not_cancelled_df_adr['reservation_status_date'] < '2017-09-01')]

# Plot the data
plt.figure(figsize=(20, 6))
plt.title('Average Daily Rate', fontsize = 30)
plt.plot(not_cancelled_df_adr['reservation_status_date'], not_cancelled_df_adr['adr'], label='Not Cancelled')
plt.plot(cancelled_df_adr['reservation_status_date'], cancelled_df_adr['adr'], label='Cancelled')
plt.legend()
plt.show()
```
